### TASK 7:HANDLING MISSING VALUES

```python
data={
    "Name": ["Niklaus", "Elijah", "Rebekah", "Hayley", "Hope"],
    "Roll": [101, 102, 103, 104, 105],
    "Marks": [85, np.nan, 78, 88,np.nan],
    "Branch": ["CS", "IT", None, "CS", "IT"]
}

df=pd.DataFrame(data)
print("Original dataframe\n",df)

#finding and changing the null values
print("Missing values:\n",df.isnull())

print("Count the missing values:\n",df.isnull().sum())


```

```python
df_removed = df.dropna()
print("\nDataset after Removing Missing Values:\n", df_removed)

df_filled = df.fillna({
    "Marks": df["Marks"].mean(),   
    "Branch": "Unknown"           
})
print("\nDataset after Filling Missing Values:\n", df_filled)

#You identify missing values with isnull(), count them, remove incomplete rows, and fill gaps using methods like mean or default values. 
#This ensures clean and reliable datasets for analysis.

#In Data Analytics, missing values can distort results and lead to wrong conclusions.
#If ignored, they may cause errors in calculations or bias in statistical models.
#Proper handling (removing or filling) ensures the dataset is clean, reliable, and ready for analysis.
```

### TASK 8:MERGE,CONCATENATE,GROUPBY & PIVOT TABLES

```python
df1 = pd.DataFrame({
    "EmpID": [101, 102, 103, 104],
    "Name": ["Riddhi", "Aman", "Priya", "Karan"],
    "Department": ["HR", "IT", "Finance", "IT"]
})

df2 = pd.DataFrame({
    "EmpID": [102, 103, 104, 105],
    "Salary": [50000, 60000, 55000, 45000],
    "Bonus": [5000, 7000, 6000, 4000]
})

print("DataFrame1:\n",df1)
print("\nDataFrame2:\n",df2)

#merging the dataframes
merged=pd.merge(df1,df2,on="EmpID",how="inner")
print("\nMerged:\n",merged)
```

```python
#concatenating the dataframes
concat=pd.concat([df1,df2],axis=0,ignore_index=True)
print("\nConcatenate:\n",concat)

#grouping the dataframes
grouped = merged.groupby("Department").agg({
    "Salary": "mean",
    "Bonus": "sum"
})
print("\nGroupBy Department (Mean Salary, Sum Bonus):\n", grouped)

#making of a pivot table
pivot = pd.pivot_table(merged, values="Salary", index="Department", columns="Name", aggfunc="mean", fill_value=0)
print("\nPivot Table:\n", pivot)
```

### TASK 10:MINI DATA ANALYSIS PROJECT

```python
import pandas as pd
import numpy as np

#load the dataset
data={
    "EmpID": [101, 102, 103, 104, 105],
    "Name": ["Niklaus", "Elijah", "Rebekah", "Hayley", "Kol"],
    "Department": ["HR", "IT", "Finance", "IT", "HR"],
    "Salary": [50000, 60000, np.nan, 55000, 52000],
    "Bonus": [5000, 7000, 6000, np.nan, 4000]
}

df=pd.DataFrame(data)
print("DataFrame\n:",df)

```

```python
#data inspection
print("\nShape:", df.shape)
print("Columns:", df.columns)
print("Data Types:\n", df.dtypes)
```

```python
print("\nInfo:")
print(df.info())
print("\nDescribe:\n", df.describe())
```

```python
#identifying and handling missing values
print("\nMissing Values:\n", df.isnull().sum())
df.fillna({"Salary": df["Salary"].mean(), "Bonus": df["Bonus"].median()}, inplace=True)
print("\nDataset after Handling Missing Values:\n", df)
```

```python
#selecting and filtering data
print("\nSelect Name & Salary:\n", df[["Name", "Salary"]])
print("\nEmployees with Salary > 55000:\n", df[df["Salary"] > 55000])
```

```python
#arraning the data
print("\nSort by Salary (Ascending):\n", df.sort_values(by="Salary"))
print("\nSort by Salary (Descending):\n", df.sort_values(by="Salary", ascending=False))
```

```python
#groupby analysis
grouped = df.groupby("Department").agg({
    "Salary": "mean",
    "Bonus": "sum"
})
print("\nGroupBy Department:\n", grouped)
```

```python
#pivot table
pivot = pd.pivot_table(df, values="Salary", index="Department", columns="Name", aggfunc="mean", fill_value=0)
print("\nPivot Table:\n", pivot)
```

```python
#insights
print("\nInsights:")
print("- Highest salary:", df["Salary"].max(), "by", df.loc[df["Salary"].idxmax(), "Name"])
print("- Average salary per department:\n", grouped["Salary"])
print("- Total bonus per department:\n", grouped["Bonus"])
```

```python
#export clean dataset
df.to_csv("cleaned_employee_data.csv", index=False)
print("\nCleaned dataset exported successfully!")
```

This mini project is a complete demonstration of how to analyze and clean data using NumPy and Pandas. We started by loading an employee dataset that contained missing values, then inspected its structure with functions like info() and describe(). Missing salaries and bonuses were handled by filling them with mean and median values to ensure consistency. Next, we selected and filtered specific records, sorted employees by salary, and used GroupBy to calculate average salaries and total bonuses per department. A Pivot Table was created to summarize salaries across departments and employees, making comparisons easier. Finally, we extracted useful insights such as the highest salary and department-level statistics, and exported the cleaned dataset to a CSV file. In short, this project shows the end-to-end workflow of data analysis — from raw data to meaningful insights and a ready-to-use dataset.
