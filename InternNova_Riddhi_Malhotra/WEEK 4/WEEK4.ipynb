### TASK 1:STATISTICS,MEAN,MEDIAN,MODE

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## We need to import statistics library to use statistics functions
from statistics import mean, median, mode

data=[10,20,30,30,40,50,50,10,10,10]

print("DATASET:",data)

#Mean → The average value, showing the central tendency of the dataset.
mean=mean(data)
print("\nMEAN: ",mean)

#Median → The middle value, less affected by extreme values/outliers.
median=median(data)
print("\nMedian: ",median)

#Mode → The most frequent value, showing the most common observation.
mode=mode(data)
print("\nMode: ",mode)

##This task introduces the three measures of central tendency — Mean, Median, and Mode.
#The mean represents the average score, giving an overall idea of performance. 
#The median identifies the middle score, which is less affected by extreme values and shows the central point of the data. 
#The mode highlights the most frequently occurring score, indicating the most common observation.
#Together, these measures provide a comprehensive understanding of the dataset’s distribution.
```

### TASK 2: VARIANCE AND STANDARD DEVIATION

```python
data=[10,20,30,30,40,50,50,10,10,10]


#Variance → Measures how far values spread from the mean.
variance=np.var(data)
print("Variance:",variance)

#Standard Deviation → Indicates the average distance of values from the mean.
standard_deviation=np.std(data)
print("\nStandard deviation:",standard_deviation)

#In Task 2, the dataset was analyzed for variance and standard deviation. 
#The variance shows the overall spread of the data, while the standard deviation indicates that, on average, each value lies about 15.72 units away from the mean.
#A smaller standard deviation would mean the dataset is tightly clustered, while a larger one would mean the dataset is more dispersed.
#These measures are crucial in statistics because they quantify data variability, which is essential for risk analysis, quality control, and decision‑making.
```

### TASK 3: CORRELATION AND PROBABILITY BASICS

### TASK 3 PART 1: CORRELATION

```python
# DATASET
age = [18, 22, 25, 30, 35, 40, 45]
spending = [2000, 2500, 2700, 3000, 3200, 3500, 3700]

# Calculate correlation coefficient
correlation = np.corrcoef(age, spending)[0,1]

print("Correlation between Age and Monthly Spending:", correlation)

# Interpretation
if correlation > 0.7:
    relation = "Strong Positive"
elif correlation > 0:
    relation = "Weak Positive"
elif correlation < -0.7:
    relation = "Strong Negative"
elif correlation < 0:
    relation = "Weak Negative"
else:
    relation = "No Correlation"

print("Relationship:", relation)

#Correlation measures the strength and direction of the relationship between two numerical variables.
# Values range from −1 to +1:
      #+1 → Perfect positive relationship (both increase together).
      #−1 → Perfect negative relationship (one increases, the other decreases).
      #0 → No relationship.

#As age increases, spending also increases. This is a strong positive correlation.      
```

### TASK 3 PART 2: PROBABILITY

```python
# Probability Example: Bag of balls
total_balls = 10
red_balls = 4

# Probability of drawing a red ball
prob_red = red_balls / total_balls

print("Probability of drawing a Red ball:", prob_red)

# Explanation
print(f"This means there is a {prob_red*100:.0f}% chance of drawing a red ball.")

#Probability measures the likelihood of an event occurring.
#Values range from 0 to 1:
      #0 → Impossible event.
      #1 → Certain event.

#Probability is used in risk analysis, forecasting, and decision-making.

```

### TASK 4:OUTLIER DETECTION

```python
data=[10,12,20,24,18,16,15,19,25,100]

mean = np.mean(data)
std_dev = np.std(data)

# --- Method 1: Z-Score ---
outliers_z = [x for x in data if abs((x - mean) / std_dev) > 2]
print("Outliers using Z-Score:", outliers_z)

#Z-SCORE:Measures how far each data point is from the mean in terms of standard deviations.

# --- Method 2: IQR ---
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = [x for x in data if x < lower_bound or x > upper_bound]

print("Outliers using IQR:", outliers_iqr)

#IQR:Uses percentiles to detect values that lie far outside the typical range.



#Outliers are unusual data points that lie far outside the typical range of values.
#They can arise due to errors, rare events, or anomalies. 
#Outliers affect analysis by skewing averages, inflating variance, and distorting correlations, which can lead to misleading conclusions.
#Detecting and handling outliers is essential to ensure that statistical results are accurate and reliable.
#In some cases, outliers may represent valuable insights (such as fraud detection or exceptional performance) 
#and should be studied carefully rather than simply removed.
```

## TASK 6: SEABORN VISUALISATION

```python
df = pd.read_csv("sample-simple.csv")
df["joined"] = pd.to_datetime(df["joined"])
```

```python
## COUNT PLOT
sns.countplot(x="city",data=df,palette="pastel")
plt.title("EMPLOYEE COUNT BY CITY")
plt.xlabel("CITY")
plt.ylabel("COUNT")
plt.show()

#Count Plot – Employee Count by City  
#Shows how many employees are in each city. Useful for understanding workforce distribution across locations.
```

```python
##BOX PLOT
sns.boxplot(x="city",y="salary",data=df,palette="Set2")
plt.title("Salary Distribution by City")
plt.xlabel("City")
plt.ylabel("Salary (₹)")
plt.show()

#Box Plot – Salary Distribution by City  
#Displays salary spread in each city, highlighting median, quartiles, and outliers. Helps compare pay levels across cities.
```

```python
##HEATMAP
corr = df[["age", "salary"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap (Age vs Salary)")
plt.show()

#Heatmap – Correlation (Age vs Salary)  
#Shows correlation values between age and salary. If close to +1, older employees earn more; if close to 0, no strong relationship.
```

```python
##PAIR PLOT
sns.pairplot(df[["age", "salary"]], diag_kind="kde")
plt.suptitle("Pair Plot of Age and Salary", y=1.02)
plt.show()

#Pair Plot – Age vs Salary  
#Displays scatter plots and distributions for age and salary. Helps visualize whether there’s a trend (e.g., higher age → higher salary).
```

In Task 6, Seaborn was used to create four visualizations: count plot, box plot, heatmap, and pair plot. These charts provide insights into employee distribution, salary variation, and relationships between age and salary. Seaborn’s advanced styling makes patterns easier to interpret compared to basic Matplotlib charts.

```python
##  data cleaning
# 1. Identify missing values
print("Missing Values:\n", df.isnull().sum())

# 2. Handle missing values (example: fill with mean for salary, fill with 'Unknown' for city)
df["salary"].fillna(df["salary"].mean(), inplace=True)
df["city"].fillna("Unknown", inplace=True)
```
